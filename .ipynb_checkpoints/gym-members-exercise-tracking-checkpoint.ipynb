```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/kaggle/input/gym-members-exercise-dataset/gym_members_exercise_tracking.csv')
df
```

```python
print(df.isna().sum())
print(df.isnull().any())
```

```python
df.info()
```

```python
df.nunique()
```

```python
plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='Workout_Type', palette='Set2')
ax.bar_label(ax.containers[0])
plt.title('Distribusi Tipe Latihan Member')
plt.show()
```

```python
edu_count = df['Workout_Type'].value_counts()

explode = [0.1] * len(edu_count) 

plt.figure(figsize=(8, 8))
plt.pie(edu_count, labels=edu_count.index, shadow=True, autopct='%1.1f%%', explode=explode, colors=['#3498db', '#e74c3c', '#2ecc71', '#ff9505'], wedgeprops={'edgecolor': 'black'})

plt.title('Distribusi Tipe Latihan Member')

plt.show()
```

```python
plt.figure(figsize=(10, 6))
ax = sns.countplot(data=df, x='Workout_Frequency (days/week)', palette='Set2')
ax.bar_label(ax.containers[0])
plt.title('Distribusi Frekuensi Latihan Member')
plt.show()
```

```python
edu_count = df['Workout_Frequency (days/week)'].value_counts()

explode = [0.1] * len(edu_count) 

plt.figure(figsize=(8, 8))
plt.pie(edu_count, labels=edu_count.index, shadow=True, autopct='%1.1f%%', explode=explode, colors=['#3498db', '#e74c3c', '#2ecc71', '#ff9505'], wedgeprops={'edgecolor': 'black'})

plt.title('Distribusi Frekuensi Latihan Member')

plt.show()
```

```python
distribution = df.groupby(['Workout_Type', 'Workout_Frequency (days/week)']).size().unstack()

ax = distribution.plot(kind='barh', stacked=True, figsize=(10, 6), colormap='Set2')

for container in ax.containers:
    ax.bar_label(container, label_type='center', fontsize=10, color='white')

plt.xlabel('Jumlah Member Gym')
plt.ylabel('Tipe Latihan')
plt.title('Distribusi Jumlah Member Berdasarkan Tipe Latihan dan Frekuensi Latihan')

plt.legend(title='Frekuensi Latihan', loc='upper right', bbox_to_anchor=(1.05, 1))

plt.tight_layout()
plt.show()
```

```python
distribution = df.groupby(['Workout_Type', 'Workout_Frequency (days/week)']).size().unstack()

plt.figure(figsize=(10, 6))

sns.lineplot(data=distribution, linewidth=2, palette='Set2', legend=False, dashes=False)

for i, frequency in enumerate(distribution.columns):
    y_val = distribution[frequency].iloc[-1]  
    plt.text(x=distribution.index[-1], y=y_val, s=frequency, color=sns.color_palette('Set2')[i])

plt.xlabel('Tipe Latihan', fontsize=12)
plt.ylabel('Jumlah Member Gym', fontsize=12)
plt.title('Distribusi Jumlah Member Berdasarkan Tipe Latihan dan Frekuensi Latihan', fontsize=12)

plt.xlim(distribution.index.min(), distribution.index.max())

plt.tight_layout() 
plt.show()
```
